# Datacleaning

```python
df.isnull().sum()
```

Insight:
The df.isnull().sum() output shows that the dataset initially had 271 missing values in the Year column and 58 missing values in the Publisher column. All other columns had zero missing values. Therefore, the missing values in Year and Publisher need to be handled before performing further data analysis.

```python
df['Year'] = df['Year'].fillna(df['Year'].median())
df['Publisher'] = df['Publisher'].fillna('Unknown')
df.isnull().sum()
```

### Insight:
The dataset contained missing values in the Year and Publisher columns. The missing Year values were replaced with the median year, while missing Publisher values were replaced with "Unknown". This helped to complete the dataset and ensure that missing values do not affect further analysis.

```python
# Check number of duplicate rows
print("Duplicate rows:", df.duplicated().sum())

#Remove duplicate rows
df = df.drop_duplicates()

#Check again
print("Duplicate rows after removal:", df.duplicated().sum())
```

### Insight

This code identifies and removes duplicate records. Removing duplicates ensures that the same video game is not counted multiple times, which improves the accuracy of statistical analysis.

```python
#Convert Year to integer
df['Year'] = df['Year'].astype(int)
```

```python
#Convert sales columns to numeric
sales_columns = ['NA_Sales', 'EU_Sales', 'JP_Sales', 
                 'Other_Sales', 'Global_Sales']

for col in sales_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
#Check data types
print(df.dtypes)
```

### Insight

This code ensures that Year is stored as an integer and sales columns are stored as numeric values. Correct data types are essential for calculations such as sum, mean, minimum, and maximum

## Univariate Analysis

###  A.Global Sales Distribution

```python
#Univariate Analysis – Global Sales Distribution
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.hist(df['Global_Sales'], bins=30)

plt.title('Distribution of Global Video Game Sales')
plt.xlabel('Global Sales (Millions)')
plt.ylabel('Number of Games')

plt.show()
```

### Insight

The histogram shows the distribution of global video game sales. Most games have relatively low sales, while a small number of games have extremely high sales. This indicates that the video game sales data is right-skewed.The distribution is right-skewed, indicating that blockbuster games are rare compared with games that achieve lower sales.

## Top 10 Platforms by Global Sales

```python
#Top 10 Platforms by Global Sales
platform_sales = (
    df.groupby('Platform')['Global_Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 5))

platform_sales.plot(kind='bar')

plt.title('Top 10 Platforms by Global Sales')
plt.xlabel('Platform')
plt.ylabel('Global Sales (Millions)')

plt.xticks(rotation=45)

plt.show()

```

### Insight

This analysis identifies the top 10 gaming platforms based on total global sales. It helps determine which platforms have had the strongest commercial performance in the video game market.The analysis shows that PS2 has the highest global sales among the top 10 platforms, followed by Xbox 360, PS3, and Wii. This indicates that these platforms had the strongest commercial performance in the video game market. Older platforms such as GBA, PSP, PS4, and PC show comparatively lower total sales in this dataset.

## Top 10 Best-Selling Games

```python
#Top 10 Best-Selling Games
top_games = df.sort_values(
    by='Global_Sales',
    ascending=False
).head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top_games['Name'],
    top_games['Global_Sales']
)

plt.title('Top 10 Best-Selling Video Games')
plt.xlabel('Global Sales (Millions)')
plt.ylabel('Game')

plt.gca().invert_yaxis()

plt.show()
```

### Insight

This visualization displays the 10 best-selling video games based on global sales. It helps identify the most commercially successful games in the dataset.The visualization shows the Top 10 best-selling video games based on global sales. Wii Sports is the best-selling game in the dataset, with significantly higher sales than the other games. Other popular titles include Super Mario Bros., Mario Kart Wii, Wii Sports Resort, and Pokémon Red/Pokémon Blue. This indicates that Nintendo games and franchises have strong commercial success in the video game market.

## B. Correlation Analysis

```python
#Correlation Analysis
#Select sales columns
sales_data = df[
    ['NA_Sales', 'EU_Sales', 'JP_Sales',
     'Other_Sales', 'Global_Sales']
]

#Calculate correlation
correlation = sales_data.corr()

print(correlation)
```

Correlation analysis shows the relationship between sales in different regions and global sales. A strong positive correlation indicates that higher sales in one region are generally associated with higher global sales.The correlation analysis shows a strong positive relationship between regional sales and global sales. North American (NA) sales have the strongest correlation with Global Sales (0.941), followed by European (EU) sales (0.903). This indicates that strong performance in North America and Europe is closely associated with higher overall global sales. Japan shows a moderate positive correlation (0.612), while Other regions show a strong positive correlation (0.748).

### 4. Regional sales measure- Donut chart

```python
###Calculate total regional sales
regional_sales = df[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].sum()

###Create donut chart
plt.figure(figsize=(7, 7))

plt.pie(
    regional_sales,
    labels=regional_sales.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops=dict(width=0.4)
)

plt.title('Regional Sales Distribution')
plt.show()
```
